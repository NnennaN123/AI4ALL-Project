```python
import joblib
import pandas as pd
import numpy as np

# Load the trained XGBoost model (using .plk extension)
xgb_model = joblib.load('xgboost/xgboost_wind_model.pkl')

# Load the dataset (assuming the dataset is in CSV format)
data = pd.read_csv('datasets/nrel_training_data.csv')

# Pick a turbine site from the dataset (assuming 'has_turbine' column exists and 1 = turbine present)
turbine_site = data[data['isTurbine'] == True].sample(1)

# Select the same features used for the model
features = ['fraction_of_usable_area', 'capacity', 'wind_speed', 'capacity_factor']
X = turbine_site[features]

```

```python
# Display results for demo
prediction = xgb_model.predict(X)
pred_proba = xgb_model.predict_proba(X)
print("Turbine Site Data:")
print(X)
print("\nModel Prediction:", "Turbine Present" if prediction[0] == 1 else "No Turbine")
print("Prediction Probability [No Turbine, Turbine]:", pred_proba[0])
```
